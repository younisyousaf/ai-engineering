## Function
A reuseable block of code that performs a specific task.

### Benefits
- Increase the resus-ability of code
- helps make programs organized

```python
def greet():
    print("Hello, World!")
    
greet()
```

```python
def greet(name):
    print("Hello", name)
    
greet("Younis")
```

```python
greet("Tauqeer")
```

```python
# Return value of function
def add(a,b):
    return a + b

result = add(3,5)
print(result)
```

```python
def square(num):
    return num * num

result = square(5)
print(result)
```

### Local vs Global Variables
1. A variable declared inside a function is called a local variable.
2. A variable declared outside the function(in the main program) is called a global variable.

```python
def show():
    x = 5 #local
    print(x)
    
x = 10 #global
show()

print(x)
```

### Lamda Function
A lambda function is a small anonymous function written in single line.

#### Usage
- for simple one-time tasks
- with built-in functions like `map()` & `filter()`

```python
add = lambda a , b: a + b
print(add(3,5))
```

```python
double_it = lambda n: n*2
print(double_it(5))
```

```python
nums = [1,2,3,4]
squares = list(map(lambda x: x*x, nums))

print(squares)
```

```python
numbers = [1,2,3,4,5,6,7,8]
odd_numbers = list(filter(lambda x: x % 2 !=0, numbers))
print(odd_numbers)
```
