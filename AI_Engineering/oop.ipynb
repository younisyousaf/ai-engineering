	Combintion of related variables & functions.

- Stores Data (attibutes).
- Perform Actions (methods).

	Data abstraction means that we do not need to show all our data and functions to the outside world.

- Promotes code reusability
- Enhances modularity (elements in form of objects)
- Improves maintainability
- Makes programs close to real-world thinking

#### Encapsulation
	In Programming, related data is encapsulated into a single object. The Functions related to those variables are also grouped together.


### Core Concepts

#### Class
	A class is a blueprint for creating objects.

- Defines attributes(data) & methods(behaviour) of an object.

#### Object
	An object is a real instance of a class.
```python
	class Car:
		def __init__(self, color):
			self.color = color

		def start(self):
			print(f"{self.color} car started")

	# First object
	my_car = Car("red")
	my_car.start()

	# Second object
	mybrother_car = Car("blue")
	mybrother_car.start()
```

Together they define what the class is and what it does.
1. Attrbiutes: data stored in the object(name, age)
2. Methods: actions that the object can perform(speak(), age())

### Instance Variables & Instance Methods

#### Instance
	An instance refers to an object created from a class.



````python
class Student:
    def __init__(self, name, age, grade):
        # instance variables
        self.name = name 
        self.age = age
        self.grade = grade
    
    # Instance Methods
    def display_info(self):
        print(f"Student Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Grade: {self.grade}")
    
    def is_eligible(self):
        if self.age >=15:
            print(self.name, "is eligible for admission.")
        else:
            print(self.name, "is not eligible for admission.")
````

````python
# Creating objects
student1 = Student("Ali", 16, "10th")
student2 = Student("Sara", 14, "8th")
````

````python
# Accessing attributes
print(student1.name)
print(student2.grade)
````

````python
# Calling methods
student1.display_info()
student1.is_eligible()
````

````python
student2.display_info()
student2.is_eligible()
````

### Encapsulation

#### Private Attributes
	Those attributes that can only be used within class's own methods.
	They cannot be used at global scale.

````python
class Student:
    def __init__(self, name, grade):
        self.__grade = grade # private attribute
student1 = Student("Ali", 90)
print(student1.__grade) # Error
````

````python
class Student:
    def __init__(self, name, grade):
        self.__grade = grade # private attribute
        
    def get_grade(self):
        return self.__grade
student1 = Student("Ali", 90)
# Accessing the private attribute using a public method
print(student1.get_grade())
````

````python
class Student:
    @property
    def grade(self):
        return self.__grade
    def __init__(self, name, grade):
        self.__grade = grade
    def get_grade(self):
        return self.__grade
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            print("Invalid grade!")
student = Student("Ali", 90)
student.set_grade(95)
student.get_grade()
````

### Getters & Setters in Python
#### Getters
	Functions used to retrieve the data attributes of a class.
#### Setters
	Functions used to assign or modify the data attributes of a class.


````python
class Student:
    @property
    def grade(self):
        return self.__grade
    def __init__(self, name, grade):
        self.__grade = grade
    # getter
    def get_grade(self):
        return self.__grade
    # setter
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            print("Invalid grade!")
student = Student("Ali", 90)
student.set_grade(95)
student.get_grade()
````

### Inheritance
	Inheritance allows a new class to reuse the properties & methods of an existing class.

````python
class Animal:
    def speak(self):
        print("Some sound")

class Dog(Animal):
    pass

dog1 = Dog()
dog1.speak()
````

````python
class Dog(Animal):
    def bark(self):
        print("Woof!")
childdog = Dog()
childdog.speak()
childdog.bark()
````

### Method Overriding
	

````python
class Animal:
    def speak(self):
        print("Some sound")

class Dog(Animal):
    # Overriding parent class method
    def speak(self):
        print("Woof! Woof!")

dog1 = Dog()
dog1.speak()
````

### Polymorphism
	It refers to the ability of a function or method to exist in multiple forms.


````python
class Dog:
    def speak(self):
        print("Woof!")
````

````python
class Cat:
    def speak(self):
        print("Meow!")
````

````python
for pet in [Dog(), Cat()]:
    pet.speak()
````

````python
class Vehicle:
    def move(self):
        print("Vehicle is moving")
````

````python
class Car(Vehicle):
    def move(self):
        print("Car is moving")
car1 = Car()
car1.move()
````

### Method Overloading
	

````python
class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c

calc1 = Calculator()
# Method Overloading: Using same function with different number of arguments
print(calc1.add(5))
print(calc1.add(5,10))
print(calc1.add(5,10,15))
````
