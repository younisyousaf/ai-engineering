```python
# Loops

for i in range(3):
    print("python is fun")
    

for char in "python":
    print(char)
```

```python
i = 4
rows = i
cols = i

matrix = []
for r in range(rows):
    row = []
    for c in range(cols):
        row.append(c+1)
    matrix.append(row)
    
for row in matrix:
    print(' '.join(map(str, row)))
    
```

```python
for level in range(1, rows + 1):
    print(' ' * (rows - level) + '*' * (2 * level -1))
```

```python
# While Loop
count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
# Nested Loop
for i in range(3):
    print("Outer Loop")
    for j in range(3):
        print("	Inner Loop")
```

```python
# Iterating over string
for char in "python":
    print(char)
    
print(
	'\n'
)
count = 0
for char in "banana":
    if char == "a":
        count += 1
print(count)
```

```python
# Looping over dictionary
student = {"name" : "Ali", "age" : 20, "grade" : "A"}

#print keys only
for key in student:
    print(key)
    
#print values
for key in student:
    print(student[key])
    
#print key value
for key in student:
    print(key, ":", student[key])
    
student_list = [
	{"name": "Ali", "age" : 20, "grade":"A"},
	{"name": "sara", "age": 22, "grade" : "B"}
]

#print filtered data
for student in student_list:
    print(f"Name: {student.get('name','')} | Grade: {student.get('grade','')}")
```
