#### Python File Modes
- Read
- Write
- Append

```python
file = open("demo.txt", "r")
print(file.read())
file.close()
```

```python
file = open("demo.txt", "w")
file.write("Hello, Welcome to the Python World!")
file.close()
```

```python
file = open("demo.txt", "a")
file.write("\nPython is Awesome!")
file.close()
```

### Context Managers
	A resource management tool that ensures setup and cleanup operations are handled automatically when executing a block of code.

1. open(): The standard file context manager for reading/writing files.
2. tempfile.TemporaryFile()/NamedTemporaryFile(): Creates temporary files that delete themselves automatically.
3. zipfile.ZipFile(): Opens .zip files for reading/writing.
4. contextlib.ExitStack: Let's you manage multiple file at once easily.
5. pathlib.Path().open(): Modern, object-oriented file opening via pathlib.

```python
# Using Context Managers (with Statement)
with open("demo.txt", "r") as file:
    data = file.read()
    print(data)
#File closed automatically here!
```

### Exception Handling
	An exception is an error that occurs during the execution of a program.

```python
with open("missing.txt", "r") as file:
    print(file.read())
```

```python
try:
    with open("missing.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found!")
```

```python
filename = "my_info.txt"
name = "GitHub Copilot"
age = 1
```

```python
# Try to read existing file; if missing, create it with name and age, then read.
try:
    with open(filename, "r") as fin:
        print(fin.read())
except FileNotFoundError:
    with open(filename, "w") as fout:
        fout.write(f"{name}\n{age}\n")
    with open(filename, "r") as fin:
        print(fin.read())
```
