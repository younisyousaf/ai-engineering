### Exception
	An exception is an event that disrupts the normal flow of the program.

### Importance of Exception Handling
	Due to an error, Python will stop the program & display an error message, instead we should handle the error & allow the program to continue running.

	The program should not crash due to errors.

#### Example of errors:
1. Dividing by zero (Exception: ZeroDivisionError)
2. File not found (Exception: FileNotFoundError)
3. Invalid user input (Exception: ValueError)

When an error occurs in your program, you inform python that the error was expected and that you are handling it.

```python
num = int(input("Enter a number: "))
print(10 / num)
```

```python
try:
    num = int(input("Enter a number: "))
    print(10 / num)
except ZeroDivisionError:
    print("You can't divide by zero!")
except ValueError:
    print("That's not a number!")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
try:
    print("Running code...")
except:
    print("Error found!")
else:
    print("No errors found!")
finally:
    print("Done!")
```

###  Raising Exception

```python
age = -3
if age < 0:
    raise ValueError("Age cannot be negative!")
```

### Creating Custom Exception

```python
class InvalidAgeError(Exception):
    pass
```

```python
age = int(input("Enter age: "))
if age < 0:
    raise InvalidAgeError(f"Age cannot be negative!, {age}")
```

```python
try:
    age = int(input("Enter age: "))
    if age < 0:
        raise InvalidAgeError(f"Age cannot be negative!, {age}")
except InvalidAgeError as e:
    print(e)
```
