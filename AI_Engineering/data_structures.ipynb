# Data Structures in Python

- Lists (Ordered, editable), it can hold multiple values
- Tuple (Ordered, fixed), 
- Set (Unique, unordered)
- Dictionary (Key value storage)

```python
# Lists
fruits = ["apple", "banana", "mango", "orange"]
print(fruits)

# Accessing elements
print(fruits[0])
print(fruits[-1])

# List Methods
# add item
fruits.append("strawberry")
print(fruits)
# Insert at index
fruits.insert(1, "grapes")
# Remove item
fruits.remove("banana")
# Remove the last item
fruits.pop()
# Sort list
fruits.sort()
print(fruits)

# Loop
for fruit in fruits:
    print(fruit)
```

# Tuple
A tuple is ordered but unchangeable collection
- Ordered
- cannot modify after the creation

```python
colors = ("red", "green", "blue")
print(colors)

# access tuple
print(colors[0])

# Tuple are immutable
# colors[1]="black"
# print(colors)

# Tuple methods
numbers = (1,2,3,4,2,5,6)
print(numbers.count(2))
print(numbers.index(1))
```

# Set
A set is unique unordered values

```python
my_set = {1,2,3,4,4,5,6,7}
print(my_set) # no duplicates

my_set.add(9)
my_set.remove(2)
print(my_set)

A = { 1,2,3 }
B = { 3,4,5,6 }
print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
```

# Dictionary
Stores the data as key:value pairs

```python
student = {
	"name" : "Ali",
	"age" : 22,
	"course" : "AI Engineering"
}
print(student)

# Access values
print(student["name"])
print(student.get("age"))

# Methods
student["age"] = 23 #update
student["city"] = "Lahore" # add
student.pop("course") # remove
print(student)

# Looping
for key, value in student.items():
    print(key, ":", value)
```
