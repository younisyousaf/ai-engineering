### Function
	A function can be used in the global context & operates independently of a class.
### Method
	A method is a functio defined inside a class.

### Special Function
	A special function defines built-in behavior automatically associated with certain operations.

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
    def __str__(self):
        return f"'{self.title}' by {self.author}"
```

```python
b = Book("1984", "George Orwell")
print(b)
```

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs
    def __len__(self):
        return len(self.songs)
    def __add__(self, other):
        return Playlist(self.songs + other.songs)
```

```python
p1 = Playlist(["Song1", "Song2"])
```

```python
p2 = Playlist(["Songs"])
```

```python
print(len(p1)) #2
```

```python
print(len(p1+p2)) #3
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __str__(self):
        return f"({self.x}, {self.y})"
    def __add__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        return Point(self.x + other.x, self.y + self.y)
```

```python
p1 = Point(2, 3)
```

```python
p2 = Point(4, 1)
```

```python
print(p1 + p2)
```
