### Series
	1D labeled array.
	Consists of index + values
	Automatically infers dtype

```python
import pandas as pd
```

```python
# Create a series
s = pd.Series([10,20,30,20,40,50], index=['a', 'b', 'c', 'd', 'e','f'])
print(s)
```

```python
# Index vs Values
print("Values:", s.values)
print("Index:", s.index)
print("Dtype:", s.dtype)
```

```python
print(s.head)
print(s.tail)
print(s.value_counts())
```
