## NumPy Indexing, Slicing & Strides

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]
            	])
```

```python
print(arr[0, 1]) # 0 row 1 col
print(arr[2, 2])
```

```python
print(arr[-1,-2]) # last row, 2nd last column
print(arr[-3, -3])
```

```python
# general form: arr[start: stop: step]
print(arr[0:2, 1:3]) # rows 0-1, cols 1-2
```

```python
print(arr[::2, ::2]) #every 2nd row & 2nd column
```

```python
slice_arr = arr[0, :2] # view (not a copy)
slice_arr[0] = 999
```

```python
print(arr[0])
```

```python
copy_arr = arr[1].copy() # independent copy
copy_arr[0] = 255
```

```python
print(arr[1])
```

## NumPy Random Numbers, Reproducibility & Performance Tips

#### Random Number generation
It produces only numbers randomly.

#### Seed
A seed is an initial values that helps detemine the sequence of random numbers

```python
# Simulate 100k coin flips
rng = np.random.default_rng(seed=45)
flips = rng.integers(0,2, size = 100000)
prob_heads = np.mean(flips)
print("Estimated probability of heads:", prob_heads)
```

```python
import time

# Python Loop
start = time.time()
squares = [i**2 for i in range(1000000)]
end = time.time()
print("Loop Time:", end - start)

# Numpy vectorized Loop
start = time.time()
arr = np.arange(1000000)
squares_np = arr**2
end = time.time()
print("Vectorized Time:", end - start)
```
